```python
import numpy as np
import matplotlib.pyplot as plt
import librosa
import os
import random
from scipy.signal import correlate

def evaluate_alignment(signal1, signal2, fs):
    """Évalue le décalage temporel entre deux signaux EMG."""
    correlation = correlate(signal1, signal2, mode='full')
    lag = np.argmax(correlation) - (len(signal1) - 1)
    time_shift = lag / fs  # Décalage en secondes
    return time_shift

# 1. Paramètres
fs = 10000  # Fréquence d'échantillonnage
folder_path = "DataBase/CHEEK/Later"
num_pairs = 1000  # Nombre de paires aléatoires à comparer

# 2. Liste des fichiers dans le dossier
files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.endswith(".wav")]

# Vérification qu'on a assez de fichiers
if len(files) < 2:
    raise ValueError("Pas assez de fichiers dans le dossier pour comparer des paires.")

# 3. Sélection de paires aléatoires
random_pairs = random.sample([(f1, f2) for f1 in files for f2 in files if f1 != f2], num_pairs)

# 4. Comparer chaque paire et calculer le time_shift moyen
time_shifts = []

for file1, file2 in random_pairs:
    # Charger les fichiers en stéréo
    audio1, _ = librosa.load(file1, sr=None, mono=False)
    audio2, _ = librosa.load(file2, sr=None, mono=False)

    # Comparer les signaux du canal 2 (indice 1)
    time_shift = evaluate_alignment(audio1[1, :], audio2[1, :], fs)
    time_shifts.append(time_shift)

    print(f"{file1} vs {file2} → Décalage : {time_shift:.3f} s")

# 5. Moyenne des décalages
mean_time_shift = np.mean(time_shifts)
print(f"\nDécalage temporel moyen sur {num_pairs} paires : {mean_time_shift:.3f} s")
```
